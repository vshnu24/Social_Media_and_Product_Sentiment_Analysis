```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report
import joblib
import pandas as pd
```

```python
data = pd.read_csv("sentiment_dataset.csv")
data.head()
```

```python
X = data[["text", "product_name", "category", "platform", "source_type", "rating"]].copy()
X["text"] = X["text"].fillna("").astype(str)
y = data["sentiment"]
```

```python
# Features are transformed by the model pipeline below.
```

```python
categorical_features = ["product_name", "category", "platform", "source_type"]

preprocessor = ColumnTransformer(
    transformers=[
        ("text", TfidfVectorizer(stop_words="english", ngram_range=(1, 2)), "text"),
        ("categorical", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("onehot", OneHotEncoder(handle_unknown="ignore"))
        ]), categorical_features),
        ("rating", SimpleImputer(strategy="median"), ["rating"])
    ]
)

model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(max_iter=1000))
])
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)
```

```python
model.fit(X_train, y_train)
```

```python
predictions = model.predict(X_test)
```

```python
print("Accuracy:", accuracy_score(y_test, predictions))
print(classification_report(y_test, predictions))
```

```python
joblib.dump(model, "sentiment_model.joblib")

print("Model trained and saved successfully!")
```
